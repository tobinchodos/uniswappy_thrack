```python
import pandas as pd
import numpy as np
from lp_analysis import load_and_preprocess
from lp_analysis_multi_pool import clamp_dfs

pd.options.plotting.backend = "plotly"
pd.options.display.max_columns=100
pd.options.display.max_rows = 50

stems = ['weth_dai',"weth_usdc",'wbtc_usdt','wbtc_weth']
lp_dfs =[]
edfs =[]
for stem in stems:

    low_fee = pd.read_parquet(f"data/lp_analysis_output_{stem}_500.parquet")
    low_fee["fee_tier"] = "low_fee"
    low_fee["tokens"] = stem
    _low_edf_full = load_and_preprocess(f"data/{stem}_500.csv", n_rows=None)
    _low_edf_full["fee_tier"] = "low_fee"
    _low_edf_full["tokens"] = stem

    high_fee = pd.read_parquet(f'data/lp_analysis_output_{stem}_3000.parquet')
    high_fee["fee_tier"] = "high_fee"
    high_fee["tokens"] = stem
    
    _high_edf_full = load_and_preprocess(f"data/{stem}_3000.csv", n_rows=None)
    _high_edf_full["fee_tier"] = 'high_fee'
    _high_edf_full["tokens"] = stem

    low_edf,high_edf = clamp_dfs([_low_edf_full, _high_edf_full])

    for df in [low_fee,high_fee]:
        df['fee_roi']+=1 # quick fix!
        df['num_liq_ops'] = df['liq_idx'].apply(len)
        df.drop(columns=['liq_val','liq_idx'],inplace=True)
    
    for i,df in enumerate([low_edf, high_edf]):

        _lp_df = low_fee if i==0 else high_fee # ADDING CONSTANTS INFO TO LP DF
        
        df["liquidity"] = df.loc[
            df.event.isin(["mint", "burn"]), "amount_lp_token"
        ].cumsum()
        
        df["volume_acc"] = np.cumsum(
            np.where(
                df.event.eq("swap") & (df.amount0 > 0), df.amount0 * df.current_price, 0
            )
            + np.where(df.event.eq("swap")
            & (df.amount1 > 0),
            df.amount1,
            0,)
        )


        
        print("-" * 10 +str(stem).upper()+ f" fee = {df.fee_rate.iloc[0]} " + "-" * 10)
        print(f"start: {df.evt_block_time.min()}, stop: {df.evt_block_time.max()}")

        num_events = df.shape[0]
        num_mints = df.event.eq('mint').sum()
        num_swaps = df.event.eq('swap').sum()
        num_burns = df.event.eq('burn').sum()
        _lp_df['num_events'] = num_events
        _lp_df['num_mints'] = num_mints
        _lp_df['num_swaps'] = num_swaps
        _lp_df["num_burns"] = num_burns
        

        print(
            f"{num_events} events: {num_mints} mints, {num_swaps} swaps, {num_burns} burns "
        )

        swap_size_0_avg = df.loc[df.event.eq('swap'),'amount0'].abs().mean()
        swap_size_0_med = df.loc[df.event.eq('swap'),'amount0'].abs().median()
        swap_size_1_avg=df.loc[df.event.eq('swap'),'amount1'].abs().mean()
        swap_size_1_med=df.loc[df.event.eq('swap'),'amount1'].abs().median()

        _lp_df['swap_size_0_avg'] = swap_size_0_avg
        _lp_df['swap_size_0_med'] = swap_size_0_med
        _lp_df['swap_size_1_avg'] = swap_size_1_avg
        _lp_df["swap_size_1_med"] = swap_size_1_med

        print(
            f"swap sizes:\n\tamount0 avg {swap_size_0_avg}, med {swap_size_0_med} \n\tamount1 avg  {swap_size_1_avg}, med {swap_size_1_med}"
        )

        df["lp_id"] = df.lp_tuple.apply(lambda x: str(x[0]))
        num_pos = df.loc[df.event.isin(['mint','burn']),'lp_tuple'].nunique()
        num_addr = df.loc[df.event.isin(['mint','burn']),'lp_id'].nunique()
        _lp_df['num_pos'] = num_pos
        _lp_df['num_addr'] = num_addr


        print(f"{num_pos} positions, {num_addr} unique addresses")
        print("\n")

        



    lp_dfs.extend([low_fee, high_fee])    
    edfs.extend([low_edf,high_edf])

edf = pd.concat(edfs, axis=0)
lp_df = pd.concat(lp_dfs, axis=0)
edf.set_index("evt_block_time",inplace=True)

```
